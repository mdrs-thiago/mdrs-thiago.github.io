# Lab 03: Perdas e Métricas

Neste lab, vamos entender como calcular o erro de um modelo e como avaliar seu desempenho usando métricas clássicas.

```python
import torch
import torch.nn as nn
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Exemplo de predições e valores reais (Classificação Binária)
y_true = torch.tensor([1, 0, 1, 1, 0])
y_pred_logits = torch.tensor([2.5, -1.2, 0.5, 3.1, -0.8]) # Saída bruta da rede
y_pred = (y_pred_logits > 0).float() # Threshold simples
```

## 1. Calculando Métricas
Use as funções do sklearn para calcular acurácia e F1-score.

```python
acc = accuracy_score(y_true.numpy(), y_pred.numpy())
f1 = f1_score(y_true.numpy(), y_pred.numpy())

print(f"Acurácia: {acc:.2f}")
print(f"F1-Score: {f1:.2f}")
```

## 2. Cross Entropy Loss
Compare o cálculo manual com o `nn.CrossEntropyLoss` do PyTorch.

```python
criterion = nn.BCEWithLogitsLoss() # Binary Cross Entropy
loss = criterion(y_pred_logits, y_true.float())
print(f"Loss: {loss.item():.4f}")
```
